```python
import numpy as np
import pandas as pd 
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.head()
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder , OrdinalEncoder
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split 
X_train ,X_test , y_train ,y_test = train_test_split(df.drop(columns=['has_covid']) , df['has_covid'] ,test_size= 0.2 , random_state= 42)
```

```python
X_train
```

Using Different imputing technique

```python
# Adding simple imputer to fever col 
si = SimpleImputer()
x_train_fever = si.fit_transform(X_train[['fever']])

# test data
x_test_fever = si.transform(X_test[['fever']])

x_train_fever.shape
```

```python
# ordinal encoding
oe = OrdinalEncoder(categories=[['Mild','Strong']])
X_train_cough = oe.fit_transform(X_train[['cough']])

X_test_cough = oe.transform(X_test[['cough']])

X_train_cough.shape
```

```python
# one hot enocodig 
ohe = OneHotEncoder(drop='first',sparse_output= False)
X_train_gender_city = ohe.fit_transform(X_train[['gender','city']])

X_test_gender_city = ohe.transform(X_test[['gender' ,'city']])

X_test_gender_city.shape
```

```python
# Extracting Age 
X_train_age = X_train.drop(columns = ['gender' , 'fever' , 'cough' , 'city']).values
X_test_age = X_test.drop(columns = ['gender' , 'fever' , 'cough' , 'city']).values
X_train_age.shape
```

```python
X_train_transformed = np.concatenate((X_train_age,x_train_fever,X_train_gender_city,X_train_cough ), axis=1)

X_test_transfromed = np.concatenate((X_test_age,x_test_fever,X_test_gender_city,X_test_cough ), axis = 1)

X_train_transformed.shape
```

Column Transformer

```python
from sklearn.compose import ColumnTransformer
```

```python
ct = ColumnTransformer(transformers=[
    ('tn1' , SimpleImputer(),['fever']),
    ('tn2' , OrdinalEncoder(categories=[["Mild","Strong"]]) , ['cough']),
    ('tn3' , OneHotEncoder(drop='first' , sparse_output= False) ,['gender' , 'city'])
] , remainder='passthrough')
```

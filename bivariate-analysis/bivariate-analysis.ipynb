```python
import numpy as np
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
titanic = pd.read_csv('train.csv')
```

```python
titanic.head()
```

```python
iris = sns.load_dataset('iris')
iris.head()
```

```python
flights = sns.load_dataset('flights')
flights.head()
```

1. Scatterplot(Numerical - Numerical)


```python
sns.scatterplot(x='total_bill', y='tip', hue='sex', style='smoker', size='size', data=tips)

# Add title and labels if needed
plt.title('Tips by Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

2. Bar plot(Numerical - Categorical)

```python
titanic.head()
```

```python
sns.barplot(x='Pclass', y='Age', hue='Sex', data=titanic)
```

3. Box Plot

```python
sns.boxplot(titanic , x= 'Sex' ,y= 'Age' ,hue= 'Survived')
```

4. Distplot(Numerical-categorical)

```python
sns.displot(data=titanic, x='Age', hue='Survived', kind='kde', col='Survived')
```

5. HeatMap (Categorical - categorical)

```python
sns.heatmap(titanic.corr(numeric_only= True) , vmin= -1 , vmax= 1 , annot= True)
```

6. ClusterMap (categorical - categorical)

```python

sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```

7. Pair plot

```python
iris.head()
```

```python
sns.pairplot(iris , hue= 'species')
```

8. Lineplot(Numerical - Numerical)

```python

sns.clustermap(flights.pivot_table(values='passengers',index='month',columns='year'))
```
